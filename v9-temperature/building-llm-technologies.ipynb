# Building LLM Technologies - Temperature

## Objective

Understand how temperature changes the probabilities used to select the next token.

Low temperature makes the model more deterministic.
High temperature makes the model more random.

```python
import math
```

```python
tokens = ["Python", "code", "pizza"]
logits = [2.0, 1.0, 0.5]

temperature = 0.5

scaled_logits = [x / temperature for x in logits]

exp_values = [math.exp(x) for x in scaled_logits]
total = sum(exp_values)

probabilities = [x / total for x in exp_values]

for token, probability in zip(tokens, probabilities):
    print(token, probability)
```

```python
tokens = ["Python", "code", "pizza"]
logits = [2.0, 1.0, 0.5]

temperature = 2.0

scaled_logits = [x / temperature for x in logits]

exp_values = [math.exp(x) for x in scaled_logits]
total = sum(exp_values)

probabilities = [x / total for x in exp_values]

for token, probability in zip(tokens, probabilities):
    print(token, probability)
```

## Notes

- Temperature changes how strongly logits affect the final probabilities.
- Low temperature makes the distribution more concentrated.
- High temperature makes the distribution more balanced.
- Temperature changes randomness, not the model's knowledge.
