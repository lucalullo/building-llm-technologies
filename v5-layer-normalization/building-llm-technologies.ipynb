# Building LLM Technologies - Layer Normalization

## Objective

Understand why Transformers normalize their internal values.

Layer normalization helps keep activations stable while data moves through the network.

## Why Normalize

Values inside a Transformer can become very large or very small.

Layer normalization rescales them to a more stable range.

```python
values = [2.0, 4.0, 6.0]

mean = sum(values) / len(values)

print("Mean:", mean)
```

```python
values = [2.0, 4.0, 6.0]
mean = 4.0

centered = [x - mean for x in values]

print(centered)
```

## Normalize the Values

After centering the values around zero, we scale them using their variance.

This keeps the values on a more stable scale.

```python
values = [2.0, 4.0, 6.0]

mean = sum(values) / len(values)

variance = sum((x - mean) ** 2 for x in values) / len(values)

std = variance ** 0.5

normalized = [(x - mean) / std for x in values]

print(normalized)
```

## Notes

- Layer normalization centers values around zero.
- It also rescales them using their variance.
- This helps keep activations on a stable scale.
- Stable activations make Transformer layers easier to train.
