# Building LLM Technologies - Positional Encoding

## Objective

Understand how an LLM knows the order of tokens in a sequence.

Positional encoding adds position information to token embeddings.

## Why Position Matters

The same tokens in a different order can have a different meaning.

Example:

"I love Python"

"Python loves I"

The model needs to know not only which tokens are present, but also their positions.

```python
tokens = ["I", "love", "Python"]

embeddings = [
    [1.0, 0.0],
    [0.0, 1.0],
    [1.0, 1.0]
]

positions = [
    [0.0, 0.0],
    [0.1, 0.1],
    [0.2, 0.2]
]
```

## Add Position Information

The positional vector is added to the token embedding.

This gives the model information about both the token and its position.

```python
encoded = []

for embedding, position in zip(embeddings, positions):
    encoded.append([
        embedding[0] + position[0],
        embedding[1] + position[1]
    ])

for row in encoded:
    print(row)
```

## Different Order

If the same tokens appear in a different order, they receive different positional information.

This helps the model distinguish different sequences.

```python
tokens = ["Python", "love", "I"]

for token, position in zip(tokens, positions):
    print(token, position)
```

## Notes

- The Transformer receives token embeddings together with positional information.
- Positional information tells the model where each token appears in the sequence.
- Different models can represent position in different ways, such as learned embeddings, sinusoidal encoding, or RoPE.
