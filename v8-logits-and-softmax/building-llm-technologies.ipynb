# Building LLM Technologies - Logits and Softmax

## Objective

Understand how an LLM turns its final output into probabilities for the next token.

Logits are raw scores. Softmax converts them into probabilities.

```python
import math
```

```python
tokens = ["Python", "code", "pizza"]

logits = [2.0, 1.0, 0.5]

exp_values = [math.exp(x) for x in logits]
total = sum(exp_values)

probabilities = [x / total for x in exp_values]

for token, probability in zip(tokens, probabilities):
    print(token, probability)
```

```python
best_index = probabilities.index(max(probabilities))
next_token = tokens[best_index]

print(next_token)
```

## Notes

- Logits are raw scores produced by the model.
- Softmax converts logits into probabilities.
- Higher logits usually produce higher probabilities.
- The next token can be selected from these probabilities.
