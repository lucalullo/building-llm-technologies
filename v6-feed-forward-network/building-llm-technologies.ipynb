# Building LLM Technologies - Feed-Forward Network

## Objective

Understand how a Transformer processes each token after attention.

A feed-forward network applies a small neural transformation to each token independently.

```python
token = [1.0, 2.0]

W1 = [
    [1.0, -1.0],
    [0.5, -1.0]
]

hidden = [
    token[0] * W1[0][0] + token[1] * W1[1][0],
    token[0] * W1[0][1] + token[1] * W1[1][1]
]

print(hidden)
```

## Activation Function

After the linear transformation, an activation function introduces non-linearity.

This allows the feed-forward network to learn more complex transformations.

```python
def relu(x):
    return max(0, x)

activated = [relu(x) for x in hidden]

print(activated)
```

## Second Linear Transformation

A second weight matrix transforms the hidden values again.

This produces the final output of the feed-forward network.

```python
W2 = [
    [0.5, 0.0],
    [0.0, 0.5]
]

output = [
    activated[0] * W2[0][0] + activated[1] * W2[1][0],
    activated[0] * W2[0][1] + activated[1] * W2[1][1]
]

print(output)
```

## Notes

- A feed-forward network processes each token independently.
- `W1` transforms the token into a hidden representation.
- ReLU removes negative values.
- `W2` transforms the hidden representation into the final output.
- Feed-forward networks add non-linear processing after attention.
