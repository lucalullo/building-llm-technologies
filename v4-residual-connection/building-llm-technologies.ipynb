# Building LLM Technologies - Residual Connection

## Objective

Understand why Transformers add the original input back to the transformed output.

Residual connections help preserve information while the data moves through the network.

## Why Residual Connections

A Transformer layer changes the input representation.

A residual connection adds the original input back to the transformed result.

This helps preserve useful information.

```python
input_vector = [1.0, 2.0]
transformed = [0.5, 0.3]

output = [
    input_vector[0] + transformed[0],
    input_vector[1] + transformed[1]
]

print(output)
```

## Notes

- A residual connection adds the original input to the transformed output.
- It helps preserve information across Transformer layers.
- The input and transformed vectors must have the same size.
- Residual connections make deep Transformers easier to train.
