# Building LLM Technologies - Top-k and Top-p Sampling

## Objective

Understand how an LLM limits the possible next tokens before sampling.

Top-k keeps only the most probable tokens.

Top-p keeps the smallest group of tokens whose total probability reaches a chosen threshold.

```python
tokens = ["Python", "code", "pizza", "cat", "music"]
probabilities = [0.40, 0.30, 0.15, 0.10, 0.05]

k = 3

pairs = list(zip(tokens, probabilities))
pairs.sort(key=lambda x: x[1], reverse=True)

top_k = pairs[:k]

print(top_k)
```

```python
tokens = ["Python", "code", "pizza", "cat", "music"]
probabilities = [0.40, 0.30, 0.15, 0.10, 0.05]

pairs = list(zip(tokens, probabilities))
pairs.sort(key=lambda x: x[1], reverse=True)

top_p = []
total = 0

for token, probability in pairs:
    top_p.append((token, probability))
    total += probability

    if total >= 0.80:
        break

print(top_p)
```

## Notes

- Top-k keeps a fixed number of the most probable tokens.
- Top-p keeps the smallest group of tokens whose total probability reaches a chosen threshold.
- Both methods reduce the number of possible next tokens.
- They are used during text generation to control randomness and quality.
