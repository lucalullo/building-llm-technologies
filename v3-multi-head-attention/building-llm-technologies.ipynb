# Building LLM Technologies - Multi-Head Attention

## Objective

Understand why Transformers use multiple attention heads.

Each head can focus on different relationships between tokens.

## Why Multiple Heads

A single attention mechanism produces one view of the relationships between tokens.

Multiple heads allow the model to learn different views in parallel.

For example, one head may focus on nearby tokens while another may focus on different relationships.

```python
tokens = ["I", "love", "Python"]

head_1 = [0.7, 0.2, 0.1]
head_2 = [0.1, 0.3, 0.6]

print("Head 1:", head_1)
print("Head 2:", head_2)
```

## Different Attention Views

Each attention head produces its own set of attention weights.

This means each head can focus on different tokens.

```python
tokens = ["I", "love", "Python"]

head_1 = [0.7, 0.2, 0.1]
head_2 = [0.1, 0.3, 0.6]

for token, w1, w2 in zip(tokens, head_1, head_2):
    print(token, "Head 1:", w1, "Head 2:", w2)
```

## Combine the Heads

The outputs of the attention heads are combined into a single representation.

This allows the Transformer to use information learned by all heads.

```python
combined = []

for w1, w2 in zip(head_1, head_2):
    combined.append([w1, w2])

for token, values in zip(tokens, combined):
    print(token, values)
```

## Notes

- Each attention head can focus on different relationships between tokens.
- Multiple heads process the same sequence in parallel.
- The outputs of the heads are combined into a richer representation.
- Multi-head attention helps the Transformer capture different types of context at the same time.
