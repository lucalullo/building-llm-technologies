# Building LLM Technologies - Causal Mask

## Objective

Understand what a causal mask is, why LLMs need it, and how it prevents the model from seeing future tokens.

```python
import math
```

```python
tokens = ["I", "love", "Python"]

mask = [
    [1, 0, 0],
    [1, 1, 0],
    [1, 1, 1]
]

for row in mask:
    print(row)
```

## Attention Scores

Attention scores measure how strongly each token is related to the others.

They are computed from Query and Key vectors.

Higher score = stronger attention.

The causal mask is applied to attention scores before softmax, so future positions receive zero attention.

```python
# Raw attention scores

scores = [
    [2, 2, 0],
    [2, 4, 2],
    [0, 2, 2]
]

masked_scores = []

for i, row in enumerate(scores):
    new_row = []

    for j, score in enumerate(row):
        if mask[i][j] == 1:
            new_row.append(score)
        else:
            new_row.append(-999)

    masked_scores.append(new_row)

for row in masked_scores:
    print(row)
```

```python
def softmax(row):
    exps = [math.exp(x) for x in row]
    total = sum(exps)
    return [x / total for x in exps]

for row in masked_scores:
    print(softmax(row))
```

## Notes

- The causal mask prevents tokens from attending to future tokens.
- It is applied to attention scores before softmax.
- Masked positions receive approximately zero attention.
- This makes autoregressive text generation possible.
