# Building LLM Technologies - Transformer Block

## Objective

Understand how the main Transformer components work together inside a single block.

A Transformer block combines attention, residual connections, normalization, and a feed-forward network.

## Transformer Block Flow

A simplified Transformer block follows this flow:

Attention → Residual Connection → Layer Normalization → Feed-Forward Network → Residual Connection → Layer Normalization

```python
token = [1.0, 2.0]

attention_output = [0.5, 0.3]

residual_1 = [
    token[0] + attention_output[0],
    token[1] + attention_output[1]
]

print(residual_1)
```

## Normalize the Representation

After the residual connection, the values are normalized before moving to the next step.

```python
values = residual_1

mean = sum(values) / len(values)
variance = sum((x - mean) ** 2 for x in values) / len(values)
std = variance ** 0.5

normalized_1 = [(x - mean) / std for x in values]

print(normalized_1)
```

## Feed-Forward Network

The normalized representation is processed by a feed-forward network.

This transforms each token independently.

```python
W1 = [
    [1.0, 1.0],
    [2.0, -1.0]
]

hidden = [
    normalized_1[0] * W1[0][0] + normalized_1[1] * W1[1][0],
    normalized_1[0] * W1[0][1] + normalized_1[1] * W1[1][1]
]

print("Before ReLU:", hidden)

hidden = [max(0, x) for x in hidden]

print("After ReLU:", hidden)
```

## Second Residual Connection

The feed-forward output is added back to the representation that entered the feed-forward network.

```python
residual_2 = [
    normalized_1[0] + hidden[0],
    normalized_1[1] + hidden[1]
]

print(residual_2)
```

## Final Normalization

The second residual output is normalized again.

This produces the final output of the Transformer block.

```python
values = residual_2

mean = sum(values) / len(values)
variance = sum((x - mean) ** 2 for x in values) / len(values)
std = variance ** 0.5

normalized_2 = [(x - mean) / std for x in values]

print(normalized_2)
```

## Notes

- A Transformer block combines attention and a feed-forward network.
- Residual connections preserve the original information.
- Layer normalization keeps values stable.
- The feed-forward network transforms each token independently.
- Transformer models stack many of these blocks.
